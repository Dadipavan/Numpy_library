<h3>Array Properties</h3>
<pre>
Numpy arrays has attributes that let you inspect their structure and contents.These are not functions(no parentheses), but properties you can access directly from the array object.</pre>

```python
#arr.shape returns a tuple representing the array's dimensions(rows, columns)
arr=np.array([[1, 2, 3], [4, 5, 6]])
print(arr.shape)
```

```python
#arr.size return the total number of elements in the array.
print(arr.size)
#arr.ndim return number of dimensions of the array.
print(arr.ndim)
#arr.dtype returns the data type of the array's elements.
print(arr.dtype)
```

```python
#arr.itemsize returns the size(in bytes) of each element in the array.
print(arr.itemsize)
```

```python
#arr.T returns the transposed array
print(arr.T)
```

```python
#arr.flatten() returns a 1D copy of the array, collapsing all dimensions.
print(arr.flatten())
```

```python
#arra.reshpe(shape) return a new array with specified shape.
arr=arr.flatten()
print(arr.reshape(2,3))
```

```python
#practice
import numpy as np

# Create a 2D array
arr = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int32)

print("Shape:", arr.shape)  


print("Size:", arr.size)  


print("Dimensions:", arr.ndim)


print("Data type:", arr.dtype)  

print("Item size (bytes):", arr.itemsize) 

# arr.T: Transpose the array
print("Transpose:\n", arr.T)  


print("Flattened:", arr.flatten())  

print("Reshaped to 3x2:\n", arr.reshape(3, 2))  
```

<h3>Array Manipulations</h3>
<pre>These numpy functins let you modify, combine, or resturcture arrays.</pre>

```python
#np.concatenate((arr1,arr2,..),axis=0)
import numpy as np
arr1 = np.array([1, 2])
arr2 = np.array([3, 4])
result=np.concatenate((arr1,arr2),axis=0)
print(result)
```

```python
#np.stack((arr1,arr2,..), axis=0) stacks array along a new axis, creating a higher-dimensional array.
result=np.stack((arr1,arr2))
print(result)
result1=np.hstack((arr1,arr2))
print("horizonatal stack:\n",result)
result=np.vstack((arr1,arr2))
print("vertical stack:\n",result)
```

```python
#np.split(array, indices_or_sections, axis=0)
arr=np.array([1, 2, 3, 4, 5, 6])
result=np.split(arr,3)
print(result)
```

```python
#np.append(array, obj, values, axis=None)
result=np.append(arr,[2,4,5],axis=0)
print(result)
```

```python
#np.insert(array, object, values, axis=None)
result=np.insert(arr, 1,12323)
print(result)
```

```python
#np.delete(array, obj, axis=None)
result=np.delete(arr, 1) #Here 1 is index
print(result) 
```

```python
#np.resize(arr, new_shape) resizes an array to a new shape, repeating or truncation elements as needed.
result=np.resize(arr,(4,4))
print(result)
```

```python
#np.repead(arr, repeats, axis=None)
result=np.repeat(arr,2)
print(result)
```

```python
#np.tile(array, reps) repeats the entire array a specific number of times.
result=np.tile(arr,2)
print(result)
```

<h3>Statistical Methods</h3>
<p>Numpy's <b>linalg</b> module provides tools for matrix operations</p>

```python
import numpy as np
arr1 = np.array([1, 2])
arr2 = np.array([3, 4])
print(np.dot(arr1,arr2))
```

```python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])
print(np.matmul(arr1,arr2))
np.dot(arr1,arr2)
```

```python
result=np.linalg.inv(arr1)
print(result)
```

```python
np.linalg.det(arr1)
```

```python
eigvals, eigvecs = np.linalg.eig(arr1)
print(f"{eigvals}\n\n{eigvecs}")
```

```python
#single value decomposition
u,s,v=np.linalg.svd(arr1)
print(u,s,v)
```

```python
arr=np.array([3,4])
np.linalg.norm(arr)
```

<h3> Broadcasting and Advanced Indexing</h3>
NumPy’s broadcasting and indexing features allow efficient operations on arrays of different shapes and precise data selection.

```python
#arr[start:stop:step, start:stop:step]  slice rows and columns of an array.
arr=np.array([[1,2,3],[4,5,6]])
result=arr[0:2,1:3]
print(result)
```

```python
#arr[condition]
result=arr[arr>2]
print(result)
```

```python
#reversing
arr[::-1]
```
